## Mini Project 1: Webcam + Live Bounding Box

- We're building a scaffold project that every detection project needs.
- A Live webcam feed with alive box, nothing smart yet.

**Core Focus:**
- Frames
- Drawing
- Video loop

```python
import cv2

cap = cv2.VideoCapture(0) # 0 is the default camera

live = True

while live:
    ret, frame = cap.read()

    if not ret:
        break

    h, w = frame.shape[:2]

    # draw a box in the center, roughly where a face might be
    cv2.rectangle(frame, (w//2 - 100, h//2 - 100), (w//2 + 100, h//2 + 100), (0, 255, 0), 2)
    cv2.putText(frame, 'Face goes here', (w//2 - 100, h//2 - 110), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 1)

    #show
    cv2.imshow('Webcam', frame)

```

### Explanation:

- *cap = cv2.VideoCapture(0):*  
    - An OpenCv class that connects to a live video source.
    - 0 -> default webcam
    - **cap** in now an object that captures that open connection.
    - As long as cap is open, you can keep pulling frames from it.

- **ret, frame = cap.read():**
    - cap.read() grabs the next available frame from the webcam and returns two things: ret, frame.
    - **ret** is a boolean that that tells you whether the read succeeded or nah.
    - **frame** the next available image. It's a NumPy, like in **LESSON 1**

- **h, w = frame.shape[:2]:**
    - This a Numpy shape, (height, width, channel)
    - [:2] slices the the first 2 which is used for h, w.

- **cv2.rectangle(frame, (w//2 - 100, h//2 - 100), (w//2 + 100, h//2 + 100), (0, 255, 0), 2):**
    - Draws a rect on the frame at the given coordinates
    - the parameters are (img, (start width, start height), (end width, end height), color, tickness px)

- **cv2.imshow('Webcam', frame)**
    - Display

- **if cv2.waitKey(1) & 0xFF == ord('q'):**
    - cv2.waitKey(1) pauses for 1 millisecond waiting for a keypress, and returns a code for whatever key was pressed (or -1 if none).
- 

cap.release() closes the connection to the webcam so other programs can use it again. cv2.destroyAllWindows() closes any OpenCV display windows. Skipping these is a common bug source — your webcam can stay "locked" by Python even after the script ends.
Why this structure matters going forward: every video-based OpenCV project — face detection, motion tracking, filters — uses this exact skeleton. The only thing that changes later is what you do to frame between reading it and displaying it. In Lesson 3, instead of drawing a fixed box, we'll run a detector on frame that finds where a face actually is and draws the box there dynamically.

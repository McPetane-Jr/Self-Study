# Face & Object Detection.

##  Haar Cascade Classifiers:
Before deep learning, OpenCV used the HAAR CASCADE method foe image detection. A pre-trained classifier that scans an image at multiple scales to find a patterns of light/dark contrasts that resembles facial features.

- OpenCV ships with the pre-trained CASCADE files (XML) for eyes, smile etc
- cv2.CascadeClassifier() loads one of these XML files.
- detectMultiScale() runs the actual detection, returning a list of rectangles (x, y, w, h) — one per detected face.
- It works on grayscale images (contrast patterns, not color, matter here)
- scaleFactor — how much the image is shrunk at each scan pass (1.1 = thorough but slower, 1.3 = faster but might miss small faces)
- minNeighbors — how many overlapping detections are needed to confirm a "real" face (higher = fewer false positives, but risks missing real faces)


````python
import cv2

def main():


    # OpenCV ships these XML files inside its install; this path finds them automatically.
    face_cascade = cv2.CascadeClassifier(
        cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
    )
    # Open the default camera (usually the first camera)
    cap = cv2.VideoCapture(0)

    if not cap.isOpened():
        print("Error: Could not open camera.")
        return
    
    live = True

    while live:

        # Capture frame-by-frame
        ret, frame = cap.read()

        if not ret:
            print("Error: Could not read frame.")
            break

        # Display the resulting frame
        gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        # blurred_frame = cv2.GaussianBlur(gray_frame, (15, 15), 0)

        # contours, _ = cv2.findContours(blurred_frame, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

        # Run detection: returns a list of (x, y, w, h) rectangles
        # Those rectangles are the bounding boxes of detected faces in the image.
        
        faces = face_cascade.detectMultiScale(
            gray_frame,
            scaleFactor=1.1,   # how much the image is scaled down at each pass
            minNeighbors=5,    # higher = stricter, fewer false positives
            minSize=(60, 60)   # ignore detections smaller than this (in pixels)
        )

        # Draw a bounding box + label for every face found
        for (x, y, w, h) in faces:
            cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)
            cv2.putText(
            frame, "Face", (x, y - 10),
            cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 0), 2
            )

        # Show a live face count in the corner
        cv2.putText(
        frame, f"Faces detected: {len(faces)}", (10, 30),
        cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 255), 2
        )
 
        cv2.imshow("Lesson 3: Face Detection", frame)
 
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break
 
    cap.release()
    cv2.destroyAllWindows()

if __name__ == "__main__":
    main()
````

## Explanation



### Initialize Haar Cascade:

    - ***haarcascade_frontalface_default.xml*** -> This is concatenated to the path for the type of pattern we want (The frontal face)

# Group Anagrams

```python
class Solution:
    def groupAnagrams(self, strs: List[str]) -> List[List[str]]:
        #Have a result hashmap
        res = defaultdict(list)

        #Check each string in strs
        for str in strs:

        #With each string, make a list of 26 zeros denoting each char's freq

            freq = [0]*26 #a...z

        #Check each char in each string
            for char in str:

        #Find the index of that char in the freq list (0 based index)
        #Increment each occurance of that char (each index)
                freq[ord(char)-ord('a')] += 1
        
        #Use that list as a key
        #But lists can't be keys, so convert to a tuple
        #Append those values in the hashmap at that certain key
            res[tuple(freq)].append(str)
        
        #Return the result hashmap's values (Not the keys)
        return list(res.values())
```

# TwoSum II

Given an array of integers **sorted in ascending order**, find two indices such that such that the values add up to the target number.

Index 1 must be less than index 2.

Note:

Returned answers **are NOT** 0-based

Example:
```
input = [1,3,4,5,7,10,11]
target = 9
output = [3, 4]
```

# Approach

Since the data is said to be sorted, we gotta utilise that. So, instead of using a hashmap to find the different between the target and the current valiue, like in the OG twoSum question, we could use pointers, stay with me now.

- We can use a left pointer to start from the lowest value
- Then a right pointer to start on the highest value in the array
- *If* the *sum* of the two values at each pointers is *greater* than the target value, we **decrement** the right pointer
- *If* the *sum* of the two values at each pointer is *less*, then we know we have to **increment** the left pointer
- We repeat this until the indices are found
- We know they don't exist when the right pointer becomes less than the left pointer

~ O(n) if we'd have to iterate through the entire list.

````python
class NeetSol:
    def twoSumII(arr:list[int], tar:int):
        leftPointer, rightPointer = 0, len(arr)-1

        while leftPointer < rightPointer:

            leftValue, rightValue = arr[leftPointer], arr[rightPointer]
            sum = (leftValue + rightValue)

            if sum < tar: leftPointer+=1

            elif sum > tar: rightPointer-=1

            else: return [1 + leftPointer, 1 + rightPointer]

        return [-1, -1]


#Test

input = [1,3,4,5,7,10,11]
tar = 9 

indices = NeetSol

print(indices.twoSumII(input, tar))
````

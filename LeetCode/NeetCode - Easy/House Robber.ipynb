# House Robber

You're a professional robber planning to robhouses along a street. Each house has a certain amount of money stashed, the only sting stopping you from robbing each of them is that the adjacent houses has security systems installed and it will automatically contact the police (useless idots) if two adjacent houses are hit on the same night.

Given a list of non-negative integers representing the ammount of money of each house, determine the maximium anount of money you can robb without alertin the police in one night.

Example:
```
stash = [1,2,3,1]
output = 4
```
